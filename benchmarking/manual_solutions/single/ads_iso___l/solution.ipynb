```python
import pandas as pd
'''
simulation_1
'<tool response name=output_parser>
Based on the RASPA simulation output:

Enthalpy of adsorption: 4485445.20452 KJ/MOL (with uncertainty of ±20.700310)

Internal energy components of the adsorbate:
- Average Adsorbate Bend angle energy: 109382317171.76231 K
- Average Adsorbate Torsion energy: 171731.29914 K
- Average Adsorbate Intra Van der Waals energy: -7150.8133 K
</tool response>'

simulation_2
<tool response name=output_parser>
Based on the RASPA simulation output:

Enthalpy of adsorption: 4485450.34011 ± 12.408541 KJ/MOL

Internal energy components of the adsorbate:
- Bend angle energy: 95216507834.44398 ± 4880988901.82590 K
- Torsion energy: 148963.87085 ± 8760.02178 K
- Intra Van der Waals energy: -6224.73062 ± 319.09234 K
</tool response>
'''

results = pd.DataFrame({
    "simulation": ["simulation_1", "simulation_2"],
    "Adsorption enthalpy (kJ/mol)": [4485445.20452, 4485450.34011],
    "internal_energy_bend_angle_k": [109382317171.76231, 95216507834.44398],
    "internal_energy_torsion_k": [171731.29914, 148963.87085],
    "internal_energy_intra_vdw_k": [-7150.8133, -6224.73062],
})
results["Internal energy (kJ/mol)"] = 1.380649e-23 * 6.02214076e23 / 1000 *(
    results["internal_energy_bend_angle_k"]
    + results["internal_energy_torsion_k"] 
    +results["internal_energy_intra_vdw_k"]
    )

results["Overall Adsorption enthalpy (kJ/mol)"] = results["Adsorption enthalpy (kJ/mol)"] - results["Internal energy (kJ/mol)"]
```
