```python
from instructions import *
import os
import sys
from contextlib import redirect_stdout, redirect_stderr
from dotenv import load_dotenv
from student.agent import RaspaAgent
load_dotenv()
```

```python
res = {}
```

```python
tasks = tasks_1
instructions = instructions_1

for i, task in tasks.items():

    path = os.path.join("output", "new_agent_testing", f"task_{i}")
    os.makedirs(path, exist_ok=True)

    ### redirect print to a file
    with open(os.path.join(path, "stdout.txt"), "w") as f:
        sys.stdout = f

        print(f"[Task {i}] {task}")
        instr = instructions[i]
        print(f"\t -> [Instruction] {instr.split(':')[0]}...")

        prompt = f"Task: \n{task}\n\nGiven: \n{instr}\n\nProvide the appropriate answer by analyzing the output."

        agent = RaspaAgent(provider="anthropic", path = path)
        agent.set_auto(True)
        reply = agent.run_with_todo_list(prompt)

        print(reply)
        res[i] = (reply, agent)


        print()
        print()
```
