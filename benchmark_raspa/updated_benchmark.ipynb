```python
run_id = "benchmark8"
training_id = "b1"
```

```python
def run_tests(name, test, prompt):

    session_id = f"{run_id}/{training_id}/{name}"
    create_session(session_id=session_id, agent_type="RASPA", provider = "anthropic")
    session = load_session(session_id)
    agent = load_agent(session)
    agent.load(f"checkpoints/{training_id}/2/")
    agent.reset_chat()
    agent.auto_run = True
    response = agent.run(prompt+test, max_iter=100)

    save_agent(session, agent)
    save_session(session_id=session_id, state = session)
    print(response)
    return agent
```
